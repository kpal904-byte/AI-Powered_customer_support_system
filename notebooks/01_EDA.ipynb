```python
data = load_all_data()

faq_df = data["faq"]
tickets_df = data["tickets"]
conversations_df = data["conversations"]

print("FAQ shape:", faq_df.shape)
print("Tickets shape:", tickets_df.shape)
print("Conversations shape:", conversations_df.shape)
```

```python
faq_df["category"].value_counts()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=faq_df,
    x="category",
    order=faq_df["category"].value_counts().index
)

plt.xticks(rotation=45)
plt.title("FAQ Category Distribution")
plt.xlabel("Category")
plt.ylabel("Number of FAQs")
plt.tight_layout()
plt.show()
```

```python
tickets_df["intent"].value_counts()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=tickets_df,
    x="intent",
    order=tickets_df["intent"].value_counts().index
)

plt.xticks(rotation=45)
plt.title("Support Ticket Intent Distribution")
plt.xlabel("Intent")
plt.ylabel("Number of Tickets")
plt.tight_layout()
plt.show()
```

```python
tickets_df["priority"].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=tickets_df,
    x="priority",
    order=["low", "medium", "high"]
)

plt.title("Support Ticket Priority Distribution")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
tickets_df["status"].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=tickets_df,
    x="status"
)

plt.title("Support Ticket Status")
plt.xlabel("Status")
plt.ylabel("Number of Tickets")
plt.show()
```

```python
conversations_df["speaker"].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=conversations_df,
    x="speaker"
)

plt.title("Customer vs Agent Messages")
plt.xlabel("Speaker")
plt.ylabel("Number of Messages")
plt.show()
```

```python
conversations_df["message_length"] = (
    conversations_df["message"].str.len()
)

conversations_df["message_length"].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    conversations_df["message_length"],
    bins=10,
    kde=True
)

plt.title("Conversation Message Length Distribution")
plt.xlabel("Message Length")
plt.ylabel("Frequency")
plt.show()
```
